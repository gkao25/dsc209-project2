# Brief EDA 
## columns

```python
df.columns
```

```python
# focusing on data columns related to co2 emissions and population
id_cols = ['Country Name', 'Country Code', 'Year']
co2_pop_cols = df.columns[df.columns.str.contains('CO2|Population')].to_list()
df2 = df[id_cols+co2_pop_cols]
df2.dropna(inplace=True, how='all')
df2
```

```python
np.sort(df2['Year'].unique())
```

```python
# filtering for year=2018
df2.groupby('Year').count().sort_values('average_value_CO2 emissions (kt)', ascending=False).loc[2018]
```

## country names
There are odd "country names" that are actually regions, continents, or transformed attributes.

Found another dataset from [this GitHub repo](https://github.com/lukes/ISO-3166-Countries-with-Regional-Codes) which has all country regional codes that match with our data. Use this to merge and filter our data Country Code. 

```python
iso = pd.read_csv('country_ISOcode.csv')
iso = iso[['name', 'alpha-3']]
iso
```

```python
df2_2018 = df2[df2['Year']==2018]
df2_2018 = df2_2018.merge(iso, how='inner', left_on='Country Code', right_on='alpha-3')
df2_2018.drop(axis=1, labels=['name', 'alpha-3'], inplace=True)
df2_2018.dropna(axis=0, inplace=True, subset='average_value_CO2 emissions (kt)')
df2_2018
```
